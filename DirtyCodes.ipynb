```python
# Identify the 12 actual Segment × Ship Mode sales columns
sales_columns = df.columns[[1, 2, 3, 4, 6, 7, 8, 9, 11, 12, 13, 14]]

sales_columns
```

```python
segment_shipmode = {
    "Consumer": ("Consumer", "First Class"),
    "Unnamed: 2": ("Consumer", "Same Day"),
    "Unnamed: 3": ("Consumer", "Second Class"),
    "Unnamed: 4": ("Consumer", "Standard Class"),
    
    "Corporate": ("Corporate", "First Class"),
    "Unnamed: 7": ("Corporate", "Same Day"),
    "Unnamed: 8": ("Corporate", "Second Class"),
    "Unnamed: 9": ("Corporate", "Standard Class"),
    
    "Home Office": ("Home Office", "First Class"),
    "Unnamed: 12": ("Home Office", "Same Day"),
    "Unnamed: 13": ("Home Office", "Second Class"),
    "Unnamed: 14": ("Home Office", "Standard Class")
}

segment_shipmode
```

```python
records = []

for column, (segment, ship_mode) in segment_shipmode.items():
    temp = df[["Segment>>", column]].copy()
    temp.columns = ["OrderID", "Sales"]
    
    temp["Segment"] = segment
    temp["Ship Mode"] = ship_mode
    
    records.append(temp)

long_df = pd.concat(records, ignore_index=True)

long_df = long_df[["Segment", "Ship Mode", "OrderID", "Sales"]]
```
